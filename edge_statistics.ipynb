```python
import argparse
import os
from collections import defaultdict
from typing import List, Tuple, Dict, Set

#from combine_connections import read_agp_file, merge_segments, reverse_segments
#from ragtag_utilities.utilities import run_oae, get_ragtag_version, get_impuT2T_version

def parse_sample_list(list_sample: str) -> list[str]:
    with open(list_sample) as f:
        return [line.strip() for line in f if line.strip()]

class EdgeAggregator:
    def __init__(self, input_path: str, samples: list[str]):
        self.input_path = input_path
        self.samples = samples
        self.edges = {}

    def aggregate(self):
        """
        edges:
        - key: (u, v) # sorted
        - value: (weight, max_score, max_sample_id)
        """
        for sample in self.samples:
            if os.path.exists(self.input_path + sample + ".info"):
                with open(self.input_path + sample + ".info", "r") as f:
                    for line in f.readlines():
                        u, v, score, dist = line.strip().split("\t")
                        pair = tuple(sorted([u, v]))
                        if pair not in self.edges:
                            self.edges[pair] = [[sample], [(int(score), int(dist))]]
                        else:
                            self.edges[pair][0].append(sample)
                            self.edges[pair][1].append((int(score), int(dist)))
            else:
                print(f"Warning: {sample} not found in {self.input_path}")
        return self.edges
```

```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import numpy as np
from sklearn.mixture import GaussianMixture

def fit_best_gmm(x, max_components=10):
    bics = []
    models = []
    x = x.reshape(-1, 1)
    for k in range(1, max_components + 1):
        gmm = GaussianMixture(n_components=k, covariance_type='full').fit(x)
        weights = gmm.weights_
        significant_components = weights > 0.1
        if significant_components.sum() < len(significant_components):
            #print(f"Model number is too large, break at k={k}")
            break
        if len(significant_components) >= 1:
            pass
            #print(f"---------------------------------- GMM ----------------------------------")
            #print(f"Number of significant components: {significant_components}")
            #print(f"Weights: {weights[significant_components]}")
            #print(f"BIC: {gmm.bic(x)}")
            #print(f"---------------------------------- GMM ----------------------------------")
        bics.append(gmm.bic(x))
        models.append(gmm)
    best_k = np.argmin(bics)
    best_model = models[best_k]
    return best_model, best_k+1
```
